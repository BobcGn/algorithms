# 实验1： 算法基础实验--数理基础与串匹配程序设计

## 伪代码
```text
BF(S, T):
    n ← S 的长度
    m ← T 的长度

    对每个候选起点 start，从 0 到 n - m（包含端点）:
        j ← 0

        当 j < m:
            如果 S[start + j] 与 T[j] 不相等:
                结束本轮内部比较
            否则:
                j ← j + 1

        如果 j == m:
            返回 start

    返回 -1
```

---

````c
#include <string.h>
#include <stdio.h>
#include <time.h>
````

````c
/**
@param s: the string to be searched
@param t: the string to be searched for
@param m: the length of the string t
@param comparations: a pointer to a variable that will store the number of character comparisons made
*/
int bf(const char s[], const char t[], int m, long long *comparations){
    int n = (int)strlen(s);
    *comparations = 0; // Initialize the comparison count

    // 特殊情况：空串匹配直接返回0
    if(m == 0){
        return 0;
    }
    
    // 遍历s的每一个位置，尝试匹配t
    for(int start = 0; start <= n - m; start++){
        // 将t的每一个字符与s的对应位置进行比较，如果有不匹配的字符，则跳出循环，继续尝试下一个位置
        for(int j = 0; j < m; j++){
            // 每执行一次比较操作，就将comparations加1
            (*comparations)++; 
            if(s[start + j] != t[j]){
                break;
            }
            if(j == m - 1){
                return start;
            }
        }
    }
    // 如果没有找到匹配的子串，则返回-1
    return -1;
}
````

````c
{
    // 修改 N 的值，观察比对次数和 CPU 时间（建议 N 为：100 ，200，400，800）
    enum { N = 100 };

    char s[N + 1];

    for (int i = 0; i < N; i++) {
        s[i] = 'a';
    }
    s[N] = '\0';

    const char t[] = "aaab";
    // const char s[] = "ababc";
    // const char t[] = "abc";
    int m = (int)strlen(t);

    long long comparisons = 0;

    // 先执行一次，获取匹配结果与单次比较次数
    int result = bf(s, t, m, &comparisons);

    printf("主串：%s\n", s);
    printf("模式串：%s\n", t);

    if (result == -1) {
        printf("查找结果：未找到模式串。\n");
    } else {
        printf("查找结果：起始下标为 %d，即第 %d 个字符。\n",
               result, result + 1);
    }

    printf("单次查找的字符比较次数：%lld\n", comparisons);

    const int repeats = 100000;
    long long measured_comparisons = 0;

    // 接收返回值，让重复调用的结果被使用
    volatile int measured_result = 0;

    clock_t begin = clock();

    for (int k = 0; k < repeats; k++) {
        measured_result = bf(s, t, m, &measured_comparisons);
    }

    clock_t end = clock();

    if (begin == (clock_t)-1 || end == (clock_t)-1) {
        printf("当前环境无法获取 CPU 时间。\n");
    } else {
        double total_seconds =
            (double)(end - begin) / CLOCKS_PER_SEC;

        printf("重复执行次数：%d\n", repeats);
        printf("总 CPU 时间：%.9f 秒\n", total_seconds);
        printf("平均每次调用 CPU 时间：%.6f 微秒\n",
               total_seconds * 1000000.0 / repeats);
    }
}
````

## 边界情况验证


````c
{
    const char *test_s[] = {
        "abc",
        "abc",
        "abc",
        "abc",
        ""
    };

    const char *test_t[] = {
        "abc",
        "c",
        "abcd",
        "",
        "a"
    };

    const int expected_result[] = {0, 2, -1, 0, -1};
    const long long expected_count[] = {3, 3, 0, 0, 0};

    for (int i = 0; i < 5; i++) {
        long long comparisons = 0;
        int m = (int)strlen(test_t[i]);

        int result = bf(test_s[i], test_t[i], m, &comparisons);

        int passed =
            result == expected_result[i] &&
            comparisons == expected_count[i];

        printf("用例 %d：S=\"%s\"，T=\"%s\"\n",
               i + 1, test_s[i], test_t[i]);

        printf("  返回值：%d，预期：%d\n",
               result, expected_result[i]);

        printf("  比较次数：%lld，预期：%lld\n",
               comparisons, expected_count[i]);

        printf("  验证结果：%s\n\n",
               passed ? "通过" : "未通过");
    }
}
````

## 结论
以字符比较作为基本操作，设主串长度为 n，模式串长度为 m，且 1 ≤ m ≤ n。BF 算法共有 n − m + 1 个候选起点，每个起点最多进行 m 次字符比较，因此最坏情况下的字符比较次数为 C(n,m) = (n − m + 1)m，一般最坏时间复杂度记为 O(nm)。本次规模实验固定模式串为“aaab”，即 m = 4，因此 C(n,4) = 4n − 12，比较次数随主串长度 n 线性增长。实验统计的比较次数与公式一致，测得的平均 CPU 时间也呈大致线性增长趋势。

边界情况：
| 主串 S | 模式串 T | 预测返回值 | 预测字符比较次数 | 原因 |
|---|---|---:|---:|---|
| `"abc"` | `"abc"` | 0 | 3 | 从起点 0 开始，三个字符全部匹配 |
| `"abc"` | `"c"` | 2 | 3 | 起点 0、1 各失败一次，起点 2 匹配成功 |
| `"abc"` | `"abcd"` | -1 | 0 | `n - m = -1`，不进入外层循环 |
| `"abc"` | `""` | 0 | 0 | 按约定，空模式串直接在起点 0 匹配成功 |
| `""` | `"a"` | -1 | 0 | `n - m = -1`，不进入外层循环 |
