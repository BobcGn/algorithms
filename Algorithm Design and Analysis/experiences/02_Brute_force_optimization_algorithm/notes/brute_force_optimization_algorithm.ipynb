## 题目1：排序问题程序设计

实验内容：输入一个待排序的序列，分别用选择排序和起泡排序两种排序方法将其变换成有序的序列，输出结果，输出时要求有文字说明。请任选一种语言编写程序实现上述算法，并分析其算法复杂度。


### 1. 选择排序

排序目标：将序列按从小到大的顺序排列。

基本思想：每一轮从尚未排序的部分找到最小元素，
将它与尚未排序部分的第一个元素交换。

伪代码：

```text
SelectionSort(A, n):
    对 i 从 0 到 n - 2:
        min_index ← i

        对 j 从 i + 1 到 n - 1:
            如果 A[j] < A[min_index]:
                min_index ← j

        如果 min_index ≠ i:
            交换 A[i] 和 A[min_index]
```


````c
#include <stdio.h>
#include <stdlib.h>
#include <time.h>
````

````c
/**
选择排序
*/
void selection_sort(int arr[], int n){
    for(int i = 0; i < n - 1; i++){
        int min_index = i;

        // 在未排序的部分中找到最小元素的索引
        for(int j = i + 1; j < n; j++){
            if(arr[j] < arr[min_index]){
                min_index = j;
            }
        }

        // 扫描结束后将最小的元素放到位置i
        if(min_index != i){
            int tmp = arr[i];
            arr[i] = arr[min_index];
            arr[min_index] = tmp;
        }
    }
}
````

````c
void print_selection_sort(int arr[], int n){

    printf("排序前：");
    for (int i = 0; i < n; i++) {
        printf("%d ", arr[i]);
    }
    printf("\n");

    selection_sort(arr, n);

    printf("选择排序后（升序）：");
    for (int i = 0; i < n; i++) {
        printf("%d ", arr[i]);
    }
    printf("\n");
}
````

````c
{
    int arr[] = {29, 10, 14, 37, 13};
    print_selection_sort(arr, sizeof(arr) / sizeof(arr[0]));
    // 顺序排列
    int arr1[] = {1, 2, 3, 4, 5};
    print_selection_sort(arr1, sizeof(arr1) / sizeof(arr1[0]));
    // 逆序排列
    int arr2[] = {5, 4, 3, 2, 1};
    print_selection_sort(arr2, sizeof(arr2) / sizeof(arr2[0]));
    // 包含重复值和负数
    int arr3[] = {3, -1, 3, 0, -1};
    print_selection_sort(arr3, sizeof(arr3) / sizeof(arr3[0]));
    // 只有一个元素
    int arr4[] = {7};
    print_selection_sort(arr4, sizeof(arr4) / sizeof(arr4[0]));
}
````

- 每轮比较次数及总比较次数公式：总比较次数为`n(n-1)/2`
- 为什么有序输入也需要完成全部比较：内层循环的执行次数只与`i`和`n`有关，和元素排列顺序无关
- 时间复杂度与额外空间复杂度：平均/最好/最坏的时间复杂度均为\(O(n^2)\)，算法只使用固定数量的辅助变量，因此额外空间复杂度为 O(1)

### 2. 冒泡排序

基本思想：从左向右依次比较相邻两个元素，
如果左边的元素大于右边的元素，就交换它们。
每轮扫描将未排序部分的最大元素移动到该部分的最右端。

伪代码（包含提前结束判断，下标从 0 开始）：

```text
BubbleSort(A, n):
    对 i 从 0 到 n - 2（包含端点）:
        swapped ← 0

        对 j 从 0 到 n - 2 - i（包含端点）:
            如果 A[j] > A[j + 1]:
                交换 A[j] 和 A[j + 1]
                swapped ← 1

        如果 swapped == 0:
            结束外层循环
```


````c
void bubble_sort(int arr[], int n){
    for (int i = 0; i < n - 1; i++) {
        int swapped = 0;  // 每轮开始时，记录为“尚未发生交换”

        // 末尾已有 i 个元素确定了最终位置
        for (int j = 0; j < n - 1 - i; j++) {

            // 升序排列：左边比右边大时交换
            if (arr[j] > arr[j + 1]) {
                int tmp = arr[j];
                arr[j] = arr[j + 1];
                arr[j + 1] = tmp;

                swapped = 1;  // 记录本轮发生过交换
            }
        }

        // 完整扫描一轮后，如果没有交换，就结束排序
        if (swapped == 0) {
            // printf("第 %d 轮没有交换，提前结束。\n", i + 1);
            break;
        }
    }
}
````

````c
void print_bubble_sort(int arr[], int n){

    printf("排序前：");
    for (int i = 0; i < n; i++) {
        printf("%d ", arr[i]);
    }
    printf("\n");

    bubble_sort(arr, n);

    printf("冒泡排序后（升序）：");
    for (int i = 0; i < n; i++) {
        printf("%d ", arr[i]);
    }
    printf("\n");
}

````

````c
{
    int arr[] = {29, 10, 14, 37, 13};
    print_bubble_sort(arr, sizeof(arr) / sizeof(arr[0]));
    // 顺序排列
    int arr1[] = {1, 2, 3, 4, 5};
    print_bubble_sort(arr1, sizeof(arr1) / sizeof(arr1[0]));
    // 逆序排列
    int arr2[] = {5, 4, 3, 2, 1};
    print_bubble_sort(arr2, sizeof(arr2) / sizeof(arr2[0]));
    // 包含重复值和负数
    int arr3[] = {3, -1, 3, 0, -1};
    print_bubble_sort(arr3, sizeof(arr3) / sizeof(arr3[0]));
    // 只有一个元素
    int arr4[] = {7};
    print_bubble_sort(arr4, sizeof(arr4) / sizeof(arr4[0]));
    // 第一轮发生交换后，数组就已经有序
    int arr5[] = {2, 1, 3, 4, 5};
    print_bubble_sort(arr5, sizeof(arr5) / sizeof(arr5[0]));
}
````

## 题目2：整数排序问题

实验内容： 任意输入 N 个整数，请用选择排序、堆排序和冒泡排序算法对输入的任意 N 个数进行有序排列，输出排序结果，并比较三种算法的时间复杂度。


### 堆排序的基本思想与伪代码

排序目标：将整数序列按从小到大的顺序排列。

基本思想：将数组看作一棵完全二叉树，先建立大顶堆，使每个父节点的值都不小于其孩子节点，最大值位于堆顶。随后反复将堆顶与当前堆的最后一个元素交换，把最大值放到最终位置；缩小堆的范围，并从根节点向下调整，使剩余部分恢复为大顶堆，直到排序完成。

下标从 0 开始，节点 i 的左、右孩子下标分别为 2 * i + 1 和 2 * i + 2。只有孩子下标小于当前堆的元素个数时，该孩子才存在。大顶堆不要求左右孩子之间有大小顺序。

向下调整时，左右子树应已分别满足大顶堆条件。比较当前节点及其存在的孩子，选出最大元素；若最大元素不在当前节点，则交换，并沿原元素下移的位置继续检查。

伪代码中所有范围均包含端点；起点已越过终点时不执行循环。n / 2 表示整数除法。SiftDown 中的 n 表示当前堆的元素个数，不包含末尾已经排好的部分。

```text
SiftDown(A, n, root):
    重复:
        largest ← root
        left ← 2 * root + 1
        right ← 2 * root + 2

        如果 left < n 且 A[left] > A[largest]:
            largest ← left

        如果 right < n 且 A[right] > A[largest]:
            largest ← right

        如果 largest == root:
            结束重复

        交换 A[root] 和 A[largest]
        root ← largest

HeapSort(A, n):
    // 第一阶段：自底向上建立大顶堆，堆大小保持为 n
    对 root 从 n / 2 - 1 递减到 0:
        SiftDown(A, n, root)

    // 第二阶段：逐个确定末尾元素，并缩小堆
    对 end 从 n - 1 递减到 1:
        交换 A[0] 和 A[end]
        SiftDown(A, end, 0)
```


````c
/**
堆排序的下沉操作
@param arr: 待调整的数组
@param n: 当前堆的元素个数（不包含已排序的部分）
@param root: 当前需要下沉的节点索引
*/
void sift_down(int arr[], int n, int root){
    while(1){
        int largest = root;
        int left = 2 * root + 1; // 左子节点索引
        int right = 2 * root + 2; // 右子节点索引

        // 比较当前节点和左孩子
        if(left < n && arr[left] > arr[largest]){
            largest = left;
        }

        // 比较当前节点和右孩子，找到三者中最大的下标
        if(right < n && arr[right] > arr[largest]){
            largest = right;
        }

        // 当前节点已经不小于两个孩子，结束调整
        if(largest == root){
            break;
        }

        int tmp = arr[root];
        arr[root] = arr[largest];
        arr[largest] = tmp;

        root = largest; // 继续向下调整
    }
}
````

````c
// 堆排序
void heap_sort(int arr[], int n) {
    // 第一阶段：从下往上建立大顶堆
    for (int root = n / 2 - 1; root >= 0; root--) {
        sift_down(arr, n, root);
    }

    // 第二阶段：将当前最大值放到末尾，再恢复剩余部分的堆
    for (int end = n - 1; end > 0; end--) {
        int tmp = arr[0];
        arr[0] = arr[end];
        arr[end] = tmp;

        // 下标 end 及之后已经排好，只调整前 end 个元素
        sift_down(arr, end, 0);
    }
}
````

````c
void print_heap_sort(int arr[], int n) {
    printf("排序前：");
    for (int i = 0; i < n; i++) {
        printf("%d ", arr[i]);
    }
    printf("\n");

    heap_sort(arr, n);

    printf("堆排序后（升序）：");
    for (int i = 0; i < n; i++) {
        printf("%d ", arr[i]);
    }
    printf("\n");
}
````

````c
{
    int arr[] = {29, 10, 14, 37, 13};
    print_heap_sort(arr, sizeof(arr) / sizeof(arr[0]));

    // 已经有序
    int arr1[] = {1, 2, 3, 4, 5};
    print_heap_sort(arr1, sizeof(arr1) / sizeof(arr1[0]));

    // 完全逆序
    int arr2[] = {5, 4, 3, 2, 1};
    print_heap_sort(arr2, sizeof(arr2) / sizeof(arr2[0]));

    // 包含重复值和负数
    int arr3[] = {3, -1, 3, 0, -1};
    print_heap_sort(arr3, sizeof(arr3) / sizeof(arr3[0]));

    // 只有一个元素
    int arr4[] = {7};
    print_heap_sort(arr4, sizeof(arr4) / sizeof(arr4[0]));
}
````

堆排序分为建堆和排序两个阶段。从最后一个非叶子节点开始向上建堆，时间复杂度为 \(O(n)\)。排序阶段进行 \(n-1\) 轮，每轮将堆顶最大值放到当前堆的末尾，并对剩余部分向下调整。单次向下调整的最坏时间复杂度为 \(O(\log n)\)，因此整个堆排序的最坏时间复杂度为 \(O(n\log n)\)。本实验使用循环实现向下调整，只需要固定数量的辅助变量，额外空间复杂度为 \(O(1)\)。

````c
// 对同一种排序重复计时，输出平均每次排序的 CPU 时间
void measure_sort(const char *name,
                  void (*sort_func)(int[], int),
                  const int original[],
                  int n,
                  int repeats) {
    int *work = malloc(n * sizeof(int));

    if (work == NULL) {
        printf("%s：内存分配失败。\n", name);
        return;
    }

    double total_seconds = 0.0;

    for (int r = 0; r < repeats; r++) {
        // 每次都恢复原始数据，复制过程不计时
        for (int i = 0; i < n; i++) {
            work[i] = original[i];
        }

        clock_t begin = clock();

        sort_func(work, n);

        clock_t end = clock();

        if (begin == (clock_t)-1 || end == (clock_t)-1) {
            printf("%s：无法获取 CPU 时间。\n", name);
            free(work);
            return;
        }

        total_seconds += (double)(end - begin) / CLOCKS_PER_SEC;

        // 当前输入是 n、n-1、……、1，正确结果应为 1、2、……、n
        // 验证过程放在计时范围之外
        for (int i = 0; i < n; i++) {
            if (work[i] != i + 1) {
                printf("%s：排序结果验证失败。\n", name);
                free(work);
                return;
            }
        }
    }

    printf("%s：n = %d，重复 %d 次，平均 CPU 时间 = %.6f 毫秒\n",
           name, n, repeats,
           total_seconds * 1000.0 / repeats);

    free(work);
}
````

````c
{
    enum { MAX_N = 4000 };

    int original[MAX_N];
    int sizes[] = {1000, 2000, 4000};
    int repeats = 5;

    for (int k = 0; k < 3; k++) {
        int n = sizes[k];

        // 构造逆序输入：n、n-1、……、1
        for (int i = 0; i < n; i++) {
            original[i] = n - i;
        }

        printf("\n逆序输入，规模 n = %d\n", n);

        measure_sort("选择排序", selection_sort,
                     original, n, repeats);

        measure_sort("冒泡排序", bubble_sort,
                     original, n, repeats);

        measure_sort("堆排序", heap_sort,
                     original, n, repeats);
    }
}
````

实验方法： 使用选择排序、带提前退出判断的冒泡排序和堆排序，对规模为 1000、2000、4000 的逆序整数序列进行排序。每种算法在每个规模上重复执行 5 次，利用 clock() 测量平均每次排序的 CPU 时间。每次排序前恢复原始数据，数组复制、结果检查和输出均位于计时范围之外。
理论分析： 选择排序始终进行 \(n(n-1)/2\) 次元素比较，时间复杂度为 \(O(n^2)\)。优化后的冒泡排序在有序输入上最好为 \(O(n)\)，但本次逆序输入无法触发无交换退出，时间复杂度为 \(O(n^2)\)。堆排序的自底向上建堆阶段为 \(O(n)\)，排序阶段为 \(O(n\log n)\)，整体最坏时间复杂度为 \(O(n\log n)\)。三个排序实现本身的额外空间复杂度均为 \(O(1)\)；测量程序另使用工作数组保存输入副本。
结果分析： 输入规模从 1000 增加到 4000 时，选择排序和冒泡排序的平均 CPU 时间分别约增加 16.17 倍和 15.40 倍，与平方级增长的预期基本一致。堆排序在三个规模上的耗时均最少，且耗时增长较缓，与其较低的理论增长量级相符。
实验结论： 本次展示的功能测试获得了正确排序结果，三组逆序数据的计时运行也未报告结果验证失败。理论分析与实测结果表明，在本实验的输入规模和运行环境下，堆排序比另外两种排序耗时更少。该结论限于本次逆序输入和测试环境，不能据此认定它在所有规模与输入排列下都最快。
