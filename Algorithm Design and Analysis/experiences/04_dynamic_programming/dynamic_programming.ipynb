## 题目1：数塔问题

实验目的：

1. 掌握动态规划法的设计思想；
2. 掌握数塔问题的具体实现过程；
3. 熟练掌握二维数组的使用方法；
4. 在掌握的基础上编程实现数塔问题的具体实现过程。

实验内容：

给出一个数塔，从该数塔的顶层出发，在每一个节点可以选择向左走或向右走，一直走到该数塔的最底层，找出一条路径，使得路径上的数值和最大，输出最大数值及其路径，输出时要求有文字说明。请任选一种语言编写程序实现上述算法，并分析其算法复杂度。


### 算法思路

用二维数组 `a` 保存数塔。约定数塔共有 `1 ≤ n ≤ 100` 层，行、列下标从 0 开始，第 `i` 行的有效列下标为 `0～i`。从节点 `a[i][j]` 出发，只能走向左孩子 `a[i + 1][j]` 或右孩子 `a[i + 1][j + 1]`。

定义 `dp[i][j]` 表示从节点 `a[i][j]` 出发，到达最底层能够得到的最大路径和，包含当前节点的数值。不同路径可能经过同一个节点，而从该节点到底层的最优结果相同，因此可以保存并复用这个子问题的答案。

最底层节点没有后续选择，其最大路径和就是节点本身的数值，因此先初始化：

```text
dp[n - 1][j] = a[n - 1][j]，其中 0 ≤ j < n。
```

对于其余节点，从当前节点到底层的路径必定先经过左右两个孩子之一。分别使用两个孩子的最大路径和，再取较大值，加上当前节点的数值，即可得到当前节点的答案：

```text
dp[i][j] = a[i][j] + max(dp[i + 1][j], dp[i + 1][j + 1])。
```

因为当前行依赖下一行的答案，所以从倒数第二行（`n - 2`）开始，逐行向上计算到第 0 行；每行只遍历列下标 `0～i`。最终，`dp[0][0]` 就是从塔顶到底层的最大路径和。

为了输出对应路径，再从塔顶自上而下遍历，用变量 `j` 记录当前列下标，初始为 0。每行输出当前节点的原始数值 `a[i][j]`；若尚未到达最底层，就比较两个孩子的 `dp` 值，选择后续路径和较大的孩子。走向左孩子时列下标不变，走向右孩子时列下标加 1；若两个值相等，固定选择右孩子，得到一条最优路径。路径上的节点数值用空格分隔。

这里用 `dp` 的值判断方向，用 `j` 记录位置，不能将最大路径和赋给列下标。最大路径和与路径输出均附带文字说明。


### 伪代码

约定输入数塔有 `1 ≤ n ≤ 100` 层，第 `i` 行包含 `i + 1` 个有效节点，所有循环均包含所写的起止下标。数塔中的数值可以为负数，递推时计算的路径和须能用 `int` 表示。

```text
TowerMaxSum(a, n):
    创建 n × n 的二维数组 dp

    // 1. 初始化最底层
    对 j 从 0 到 n - 1:
        dp[n - 1][j] ← a[n - 1][j]

    // 2. 从倒数第二层开始，向上计算最大路径和
    对 i 从 n - 2 递减到 0:
        对 j 从 0 到 i:
            dp[i][j] ← a[i][j]
                        + max(dp[i + 1][j], dp[i + 1][j + 1])

    // 3. 输出最大路径和
    输出 "最大路径和为: ", dp[0][0]，然后换行

    // 4. 输出一条最优路径
    输出 "最优路径为: "（不换行）
    i ← 0
    j ← 0

    当 i < n - 1 时:
        输出 a[i][j] 和一个空格（不换行）

        如果 dp[i + 1][j] <= dp[i + 1][j + 1]:
            j ← j + 1    // 右孩子更优或平局时，向右走

        i ← i + 1

    输出 a[n - 1][j]，然后换行
```

当 `n = 1` 时，向上递推的外层循环不执行；唯一节点的数值既是最大路径和，也是完整路径。路径恢复中只在尚未到达最底层时访问孩子，避免越过数塔底部。


### 源代码实现


````c
int max(int a, int b) {
    return (a > b) ? a : b;
}

void tower_max_sum(int a[100][100], int n) {
    int dp[100][100];

    // 1. 初始化最底层
    for(int j = 0; j < n; j++){
        dp[n-1][j] = a[n-1][j];
    }


    // 2. 从倒数第二层开始，向上计算最大路径和
    for(int i = n - 2; i >= 0; i--){
        for(int j = 0; j <= i; j++){
            dp[i][j] = a[i][j] + max(dp[i+1][j], dp[i+1][j+1]);
        }
    }
    
    // 3. 输出最大路径和，并给出文字说明
    printf("最大路径和为: %d\n", dp[0][0]);

    // 4. 从塔顶开始，输出一条对应的最优路径
    int i = 0, j = 0;
    printf("最优路径为: ");
    while(i < n - 1){
        printf("%d ", a[i][j]);
        if(dp[i+1][j] > dp[i+1][j+1]){
            i++;
        } else {
            i++;
            j++;
        }
    }
    printf("%d\n", a[n-1][j]);
}
````

### 复杂度分析

以数塔的层数 `n` 为问题规模。按通常的单位代价模型，将一次整数比较、加法、赋值及单个整数的输出视为常数时间，并假设递推中计算的路径和不超出 `int` 的表示范围。

**时间复杂度**

1. 初始化最底层时，循环执行 `n` 次，每次进行一次赋值，因此耗时为 Θ(n)。
2. 自底向上递推时，第 `i` 行需要计算 `i + 1` 个状态。每个状态只进行固定次数的数组访问、比较、加法和赋值，耗时为常数。所有非底层状态的计算次数为：

   $$
   \sum_{i=0}^{n-2}(i+1)
   =1+2+\cdots+(n-1)
   =\frac{n(n-1)}{2}.
   $$

   因此，递推阶段的时间复杂度为 Θ(n²)。
3. 恢复并输出路径时，`while` 循环执行 `n - 1` 次，最后再输出最底层节点，共输出 `n` 个节点，每次选择只进行常数次操作，因此耗时为 Θ(n)。输出最大路径和只需常数时间。

三个阶段依次执行，整体时间复杂度为：

$$
T(n)=\Theta(n)+\Theta(n^2)+\Theta(n)=\Theta(n^2).
$$

通常可写为 **O(n²)**。循环次数只取决于数塔层数，不取决于节点数值或最优路径的位置，因此最好、最坏情况下的时间复杂度均为 Θ(n²)。

**空间复杂度**

主要的辅助空间是保存各节点最大路径和的二维数组 `dp`，该数组还供后续恢复路径使用。循环变量等只占常数空间，`max` 函数没有递归，路径也直接输出，没有另外保存到数组中。函数参数 `a` 访问原数塔，不会复制整个二维数组。

按可随问题规模扩展的算法设计分析，若将 `dp` 分配为 `n × n`，需要保存 `n²` 个整数，因此辅助空间复杂度为 **Θ(n²)**，通常写为 **O(n²)**。即使只存储三角形中的有效状态，数量为 `n(n + 1) / 2`，仍为平方级。

当前源代码采用固定容量的 `dp[100][100]`，始终分配 10000 个整数，所以在支持的 `1 ≤ n ≤ 100` 范围内，实际额外存储量固定，相对于 `n` 可记为 **O(1)**；这是固定容量实现的存储特点，不表示可扩展的二维 DP 算法只需常数空间。

## 题目2：多源点最短路径问题

实验目的：

1. 掌握动态规划法的设计思想；
2. 掌握多源点最短路径问题的具体实现过程；
3. 通过这个实例进一步掌握动态规划法的运用；
4. 在掌握的基础上编程实现多源点最短路径问题的具体实现过程。

实验内容：

给定带权有向图 G = (V, E)，对任意两个顶点 u、v ∈ V，求出顶点 u 到顶点 v 的最短路径长度，输出结果，输出时要求有文字说明。请任选一种语言编写程序实现上述算法，并分析其算法复杂度。


### 算法思路

采用 Floyd–Warshall（简称 Floyd）算法，求带权有向图中任意两个顶点之间的最短路径长度。约定共有 `1 ≤ n ≤ 100` 个顶点，编号为 `0～n - 1`，图中不存在负权回路。允许边权为负，但如果存在负权回路，部分顶点对可能没有有限的最短距离，不属于此处算法说明的输入范围。

**图的表示与初始化**

用邻接矩阵 `a` 表示图：对角线元素为 0，表示从顶点自身出发、不移动的距离；存在边 `i → j` 时保存边权，多条直接边取最小权值。输入保留 `INF = 1000000000` 表示无直接边，其余 `int` 值可表示边权，包括 0。

创建 `long long` 距离矩阵 `dist`。初始化时，将输入的 `INF` 转为内部标记 `LLONG_MAX`，其余元素保存为有限距离。此时尚未允许使用任何中转点，初始距离只考虑不移动和直接边。

在本实验的 32 位 `int` 环境中，有限最短路径可以取为至多 `n - 1` 条边的简单路径。两段距离的候选和的绝对值至多为 `2 × 99 × 2³¹`，在 `long long` 的表示范围内，且与内部标记有足够间隔。

**动态规划状态与转移**

将“允许哪些顶点作为中转点”作为动态规划的阶段。中转点指路径上除起点、终点外的内部顶点。

定义 $D^{(k)}[i][j]$ 表示只允许编号 `0～k` 的顶点作为中转点时，从 `i` 到 `j` 的最短距离。初始状态 $D^{(-1)}$ 不允许任何中转点，就是初始化后的距离矩阵。

当本轮新允许顶点 `k` 作为中转点时，对每个起点 `i` 和终点 `j` 比较两种方案：

1. 不经过中转点 `k`，保留前一阶段的最短距离；
2. 经过中转点 `k`，将前一阶段的 `i → k` 和 `k → j` 两段最短路径连接起来。

两段都可达时，状态转移为：

$$
D^{(k)}[i][j]
=
\min\left(
D^{(k-1)}[i][j],
D^{(k-1)}[i][k]+D^{(k-1)}[k][j]
\right).
$$

只有两段都不同于内部不可达标记时才计算候选和。候选距离可以等于或大于输入中的 `INF`，仍作为有限值参与比较。

**计算顺序与原地更新**

使用三层循环：最外层 `k` 从 0 遍历到 `n - 1`，逐个增加允许的中转点；内层 `i`、`j` 分别遍历所有起点和终点。每轮结束后，`dist` 保存允许使用 `0～k` 作为中转点时的最短距离。

实际实现可以只使用一个二维数组 `dist` 原地更新。在不存在负权回路、对角线距离为 0 的前提下，本轮经由 `k` 的更新不会改变第 `k` 行和第 `k` 列的距离，因此可以直接使用 `dist[i][k] + dist[k][j]` 构造候选距离。

有向图中 `i → j` 与 `j → i` 是不同问题，不能只处理距离矩阵的上三角或下三角，每轮都要检查全部 `n × n` 个顶点对。

**结果输出**

所有中转点都被考虑后，按矩阵输出各顶点对的最短路径长度。行表示起点，列表示终点，编号从 0 开始。内部标记 `LLONG_MAX` 显示为 `INF`，其余位置输出数值。


### 伪代码

输入为邻接矩阵 `a`，顶点数 `1 ≤ n ≤ 100`，编号为 `0～n - 1`。对角线为 0，输入中的 `INF` 表示无直接边，图中不存在负权回路。所有循环均包含起止下标。

```text
Floyd(a, n):
    创建 n × n 的 long long 距离矩阵 dist
    unreachable ← LLONG_MAX

    // 1. 初始化距离矩阵
    对 i 从 0 到 n - 1:
        对 j 从 0 到 n - 1:
            如果 a[i][j] == INF:
                dist[i][j] ← unreachable
            否则:
                dist[i][j] ← a[i][j]

    // 2. 逐个允许顶点作为中转点
    对 k 从 0 到 n - 1:
        对 i 从 0 到 n - 1:
            对 j 从 0 到 n - 1:
                如果 dist[i][k] != unreachable 且 dist[k][j] != unreachable:
                    candidate ← dist[i][k] + dist[k][j]
                    如果 candidate < dist[i][j]:
                        dist[i][j] ← candidate

    // 3. 输出带行列说明的距离矩阵
    输出 "行表示起点，列表示终点，顶点编号从 0 开始；INF 表示不可达。"

    对 i 从 0 到 n - 1:
        对 j 从 0 到 n - 1:
            如果 dist[i][j] == unreachable:
                输出 "INF "（不换行）
            否则:
                输出 dist[i][j] 和一个空格（不换行）
        输出换行
```

`k` 对应动态规划阶段，必须放在最外层。每个阶段都遍历全部有序顶点对。输入中的 `INF` 只用于初始化，递推和输出均依据内部的 `unreachable` 判断可达性。


### 源代码实现

````c
#define INF 1000000000  // 输入邻接矩阵中的无直接边标记

void floyd(int a[100][100], int n) {
    const long long unreachable = LLONG_MAX;
    long long dist[100][100];

    // 1. 将输入标记转换为内部不可达标记
    for (int i = 0; i < n; i++) {
        for (int j = 0; j < n; j++) {
            dist[i][j] = (a[i][j] == INF) ? unreachable : a[i][j];
        }
    }

    // 2. 依次允许各顶点作为中转点，更新距离
    for (int k = 0; k < n; k++) {
        for (int i = 0; i < n; i++) {
            for (int j = 0; j < n; j++) {
                if (dist[i][k] != unreachable && dist[k][j] != unreachable) {
                    long long candidate = dist[i][k] + dist[k][j];
                    if (candidate < dist[i][j]) {
                        dist[i][j] = candidate;
                    }
                }
            }
        }
    }

    // 3. 输出距离矩阵
    printf("行表示起点，列表示终点，顶点编号从 0 开始；INF 表示不可达。\n");
    for (int i = 0; i < n; i++) {
        for (int j = 0; j < n; j++) {
            if (dist[i][j] == unreachable) {
                printf("INF ");
            } else {
                printf("%lld ", dist[i][j]);
            }
        }
        printf("\n");
    }
}
````

### 复杂度分析

以图的顶点数 `n` 为问题规模，按通常的单位代价模型，将一次整数比较、加法、赋值以及单个整数或 `INF` 的输出视为常数时间。图中不存在负权回路，有限距离和候选和使用 `long long`。

**时间复杂度**

1. 初始化距离矩阵时，两层循环分别遍历 `n` 个起点和 `n` 个终点，共执行 `n²` 次赋值，因此耗时为 Θ(n²)。
2. 动态规划递推时，最外层枚举 `n` 个中转点 `k`，每轮检查全部 `n²` 个起点、终点组合。三个循环的范围均为 `0～n - 1`，因此总共检查：

   $$
   \sum_{k=0}^{n-1}\sum_{i=0}^{n-1}\sum_{j=0}^{n-1}1
   =n\times n\times n=n^3
   $$

   个组合。每个组合只涉及常数次可达性判断、数组访问、加法、比较及赋值，所以递推阶段的时间复杂度为 Θ(n³)。
3. 输出阶段遍历整个距离矩阵，共输出 `n²` 个结果，因此耗时为 Θ(n²)。

三个阶段依次执行，整体时间复杂度为：

$$
T(n)=\Theta(n^2)+\Theta(n^3)+\Theta(n^2)=\Theta(n^3).
$$

通常可写为 **O(n³)**。即使某些路径不可达，也只是跳过候选距离的计算与更新，三个循环仍然照常执行。因此，对于当前实现，无论边的数量、权值或可达关系如何，最好、最坏情况下的时间复杂度均为 Θ(n³)。

**空间复杂度**

主要的辅助空间是距离矩阵 `dist`。算法直接在该矩阵中更新，不为每个中转点另建一张表，所以不需要存储 `n` 份距离矩阵。循环变量等只占常数空间，算法没有递归，也没有额外保存路径序列。函数参数 `a` 访问原邻接矩阵，不会因参数传递而复制整个二维数组；输入图的存储不计入辅助空间。

按可随问题规模扩展的算法设计分析，若为 `n` 个顶点分配 `n × n` 的 `dist` 数组，需要保存 `n²` 个距离，因此辅助空间复杂度为 **Θ(n²)**，通常写为 **O(n²)**。

当前源代码使用固定容量的 `dist[100][100]`，始终分配 10000 个 `long long`。因此，在支持的 `1 ≤ n ≤ 100` 范围内，实际额外存储量固定，相对于 `n` 可记为 **O(1)**。这是固定容量实现的存储特点；按顶点数扩展距离矩阵时，辅助空间仍为平方级。

## 题目3：最长公共子序列问题（选做）

实验目的：

强化学生利用动态规划算法求解实际问题的能力。

实验内容：

一个给定序列的子序列是在该序列中删除若干元素后得到的序列。给定两个序列 X 和 Y，当另一序列 Z 既是 X 的子序列又是 Y 的子序列时，称 Z 是序列 X 和 Y 的公共子序列。给定两个序列 X = {x₁, x₂, …, xₘ} 和 Y = {y₁, y₂, …, yₙ}，找出 X 和 Y 的最长公共子序列。


### 算法思路

设序列 `X` 和 `Y` 的长度分别为 `m` 和 `n`。公共子序列可以跳过元素，但保留的元素必须在两个序列中分别保持原有顺序。先用动态规划计算最长公共子序列的长度，再利用状态表恢复一条达到该长度的序列。

定义 `dp[i][j]` 为 `X` 的前 `i` 个元素与 `Y` 的前 `j` 个元素的最长公共子序列长度。状态下标表示前缀长度，非空前缀的末尾元素分别是 `X[i - 1]` 和 `Y[j - 1]`。任一前缀为空时，公共子序列只能为空，因此第 0 行和第 0 列都初始化为 0。

对于两个非空前缀，按末尾元素是否相等进行转移。

1. 若 `X[i - 1] == Y[j - 1]`，可以在较短前缀的一条最长公共子序列后接上这个相同元素，得到长度 `dp[i - 1][j - 1] + 1`。反过来，任意一条非空公共子序列删去末尾元素后，其余元素都位于两段较短的前缀中，因此原长度至多为 `dp[i - 1][j - 1] + 1`。取该值作为 `dp[i][j]`。
2. 若 `X[i - 1] != Y[j - 1]`，两个末尾元素无法相互匹配，公共子序列至少要舍去其中一个。比较舍去 `X` 末尾后的 `dp[i - 1][j]` 和舍去 `Y` 末尾后的 `dp[i][j - 1]`，取较大值。

按行从左到右填表。计算当前格时，其上方、左方和左上方的状态均已求出。填表结束后，`dp[m][n]` 为所求长度。

从 `(m, n)` 开始回溯。末尾元素相等时，当前的最长长度为 `dp[i][j]`，将该元素直接写入 `result[dp[i][j] - 1]`，然后同时缩短两个前缀。末尾元素不同时，上方状态严格大于左方状态才向上，否则向左，因此平局时固定向左。任一前缀为空时停止。回溯从后往前写入结果的对应位置，最后按下标从小到大输出前 `dp[m][n]` 个字符。

例如，`X = ABC`、`Y = BAC` 时，从 `(3, 3)` 将 `C` 写入 `result[1]` 并移至 `(2, 2)`。上方和左方的值都为 1，按约定向左移至 `(2, 1)`，将匹配的 `B` 写入 `result[0]`。到达 `(1, 0)` 后停止，输出 `BC`，长度为 2。

C 实现约定两个字符串各最多包含 100 个字符。`dp[101][101]` 包含空前缀到长度为 100 的前缀，`result[101]` 可容纳最长 100 个结果字符。输出循环按结果长度逐字符读取。


### 伪代码

设输入序列 `X`、`Y` 各最多包含 100 个字符，可为空，元素下标从 0 开始。循环的起止下标均包含在遍历范围内。

```text
LCS(X, Y):
    m ← X 的长度
    n ← Y 的长度
    创建 (m + 1) × (n + 1) 的二维数组 dp

    // 1. 初始化空前缀的状态
    对 i 从 0 到 m:
        dp[i][0] ← 0
    对 j 从 0 到 n:
        dp[0][j] ← 0

    // 2. 按行计算各对前缀的 LCS 长度
    对 i 从 1 到 m:
        对 j 从 1 到 n:
            如果 X[i - 1] == Y[j - 1]:
                dp[i][j] ← dp[i - 1][j - 1] + 1
            否则:
                dp[i][j] ← max(dp[i - 1][j], dp[i][j - 1])

    // 3. 从右下角回溯，直接写入结果的对应位置
    i ← m
    j ← n
    L ← dp[m][n]
    创建可容纳 L 个元素的数组 result

    当 i > 0 且 j > 0 时:
        如果 X[i - 1] == Y[j - 1]:
            result[dp[i][j] - 1] ← X[i - 1]
            i ← i - 1
            j ← j - 1
        否则如果 dp[i - 1][j] > dp[i][j - 1]:
            i ← i - 1
        否则:
            j ← j - 1    // 平局向左

    // 4. 按正序输出
    输出 "LCS 长度为: ", L，然后换行
    输出 "LCS 为: "（不换行）
    对 k 从 0 到 L - 1:
        输出 result[k]（不换行）
    输出换行
```

当任一输入为空时，填表没有非空状态需要更新，回溯执行零次，输出长度 0 和空序列。两个非空输入没有共同元素时，输出长度也为 0。


### 源代码实现


````c
void lcs(const char X[], const char Y[]) {
    int m = (int)strlen(X);
    int n = (int)strlen(Y);
    int dp[101][101];

    // 1. 初始化第 0 列和第 0 行
    for(int i = 0; i <= m; i++){
        dp[i][0] = 0;
    }
    for(int j = 0; j <= n; j++){
        dp[0][j] = 0;
    }

    // 2. 按行填表，计算两个非空前缀的 LCS 长度
    for(int i = 1; i <= m; i++){
        for(int j = 1; j <= n; j++){
            if(X[i-1] == Y[j-1]){
                dp[i][j] = dp[i-1][j-1] + 1;
            } else {
                dp[i][j] = max(dp[i-1][j], dp[i][j-1]);
            }
        }
    }
    // 3. 填表完成后，输出 dp[m][n] 检查结果
    int i = m, j = n;
    int lcs_length = dp[m][n];
    char result[101];
    while(i > 0 && j > 0){
        if(X[i-1] == Y[j-1]){
            result[dp[i][j]-1] = X[i-1];
            i--;
            j--;
        } else if(dp[i-1][j] > dp[i][j-1]){
            i--;
        } else {
            j--;
        }
    }

    printf("LCS 长度为: %d\n", lcs_length);
    printf("LCS 为: ");
    for(int k = 0; k < lcs_length; k++){
        printf("%c", result[k]);
    }
    printf("\n");
}
````

### 复杂度分析

设 `m`、`n` 分别为两个输入字符串的长度，`L = dp[m][n]` 为最长公共子序列的长度。按单位代价模型，每次字符比较、数组访问、整数运算和单个字符输出都视为常数时间。

**时间复杂度**

1. `strlen` 获取两个输入的长度，耗时为 Θ(m+n+1)。初始化第 `0` 列和第 `0` 行分别执行 `m + 1` 次和 `n + 1` 次赋值，共 `m + n + 2` 次。
2. 填表时，外层遍历 `m` 个非空前缀，内层对每个外层状态遍历 `n` 个非空前缀。每个格子进行常数次比较和赋值，共更新 `m × n` 个格子。
3. 回溯从 `(m, n)` 开始，每次迭代使 `i + j` 至少减 `1`，最多执行 `m + n` 步。每个匹配字符直接写入 `result[dp[i][j] - 1]`。随后输出 `L` 个字符，且 `L ≤ min(m, n)`。

填表与边界初始化给出了整体耗时的下界。取长度、回溯和输出的耗时也包含在这个数量级内，因此整体时间复杂度为

$$
T(m,n)=\Theta(mn+m+n+1)=\Theta\bigl((m+1)(n+1)\bigr).
$$

当 `m, n ≥ 1` 时，`mn` 项主导，时间复杂度为 **Θ(mn)**，通常写作 **O(mn)**。若 `m = 0`，填表和回溯均执行零次，读取 `Y` 的长度和初始化第 `0` 行使总时间为 Θ(n+1)。`n = 0` 时同理，总时间为 Θ(m+1)。

**空间复杂度**

按可随输入规模扩展的算法设计分析，`dp` 需要保存 `(m + 1) × (n + 1)` 个整数，按长度输出时，回溯结果缓冲区最多需要 `min(m, n)` 个字符，其余局部变量只占常数空间。以输入字符串以外的存储计，辅助空间为 **Θ((m+1)(n+1))**。当两个输入均非空时，可简写为 **Θ(mn)**，通常写作 **O(mn)**。

当前 C 实现的 `dp[101][101]` 和 `result[101]` 分别固定分配 10201 个 `int` 和 101 个 `char`。在支持的 `0 ≤ m, n ≤ 100` 范围内，这两块数组的分配量相对于 `m`、`n` 固定。
